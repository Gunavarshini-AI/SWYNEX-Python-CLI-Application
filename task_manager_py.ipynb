```python
%%writefile task_manager.py

def add_task(tasks):
    task = input("Enter task name: ").strip()

    if task == "":
        print("Task cannot be empty.")
    else:
        tasks.append(task)
        print("Task added successfully!")


def view_tasks(tasks):
    if not tasks:
        print("No tasks available.")
    else:
        for i, task in enumerate(tasks, start=1):
            print(i, task)


def delete_task(tasks):
    view_tasks(tasks)

    try:
        task_id = int(input("Enter task number: "))

        if 1 <= task_id <= len(tasks):
            tasks.pop(task_id - 1)
            print("Task deleted successfully!")
        else:
            print("Invalid task number.")

    except ValueError:
        print("Enter a valid number.")
```

```python

%%writefile main.py

from task_manager import add_task, view_tasks, delete_task

def main():
    tasks = []

    while True:
        print("\n===== TASK MANAGER =====")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Delete Task")
        print("4. Exit")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                add_task(tasks)

            elif choice == 2:
                view_tasks(tasks)

            elif choice == 3:
                delete_task(tasks)

            elif choice == 4:
                print("Thank you!")
                break

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter a number.")

if __name__ == "__main__":
    main()
```
